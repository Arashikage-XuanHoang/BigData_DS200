```python
import pandas as pd

# ========================
# LOAD DATA
# ========================
ratings1 = pd.read_csv("ratings_1.txt", header=None,
                       names=["UserID", "MovieID", "Rating", "Timestamp"])

ratings2 = pd.read_csv("ratings_2.txt", header=None,
                       names=["UserID", "MovieID", "Rating", "Timestamp"])

users = pd.read_csv("users.txt", header=None,
                    names=["UserID", "Gender", "Age", "Occupation", "Zip"])

movies = pd.read_csv("movies.txt", header=None,
                     names=["MovieID", "Title", "Genres"])

print(ratings1)
print(ratings2)
print(movies)
print(users)

# ========================
# MERGE RATINGS
# ========================
ratings = pd.concat([ratings1, ratings2])

# ========================
# JOIN ratings + users
# ========================
df = ratings.merge(users, on="UserID")

# ========================
# JOIN với movies
# ========================
df = df.merge(movies, on="MovieID")

# ========================
# TÍNH AVG THEO GENDER
# ========================
result = df.groupby(["Title", "Gender"])["Rating"].mean().unstack()

# ========================
# FORMAT OUTPUT
# ========================
result = result.fillna(0)

result["Male"] = result.get("M", 0)
result["Female"] = result.get("F", 0)

result = result[["Male", "Female"]]

# ========================
# IN OUTPUT
# ========================
for title, row in result.iterrows():
    print(f"{title}: Male: {row['Male']:.2f}, Female: {row['Female']:.2f}")
```

```python
import pandas as pd

ratings1 = pd.read_csv("ratings_1.txt", header=None,
                       names=["UserID", "MovieID", "Rating", "Timestamp"],
                       skipinitialspace=True)

ratings2 = pd.read_csv("ratings_2.txt", header=None,
                       names=["UserID", "MovieID", "Rating", "Timestamp"],
                       skipinitialspace=True)

users = pd.read_csv("users.txt", header=None,
                    names=["UserID", "Gender", "Age", "Occupation", "Zip"],
                    skipinitialspace=True)

movies = pd.read_csv("movies.txt", header=None,
                     names=["MovieID", "Title", "Genres"],
                     skipinitialspace=True)

# ép kiểu
for df in [ratings1, ratings2]:
    df["UserID"] = df["UserID"].astype(int)
    df["MovieID"] = df["MovieID"].astype(int)

users["UserID"] = users["UserID"].astype(int)
movies["MovieID"] = movies["MovieID"].astype(int)

# merge
df = pd.concat([ratings1, ratings2]) \
        .merge(users, on="UserID") \
        .merge(movies, on="MovieID")

# tính avg
result = df.pivot_table(index="Title",
                        columns="Gender",
                        values="Rating",
                        aggfunc="mean").fillna(0)

# in
for title, row in result.iterrows():
    print(f"{title}: Male: {row.get('M',0):.2f}, Female: {row.get('F',0):.2f}")
```
